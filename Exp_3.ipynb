```python
import pandas as pd
from sklearn.datasets import load_iris
from scipy import stats
from scipy.stats import gmean, hmean

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)

print("First Five Records")
print(df.head())
```

Mean

```python
print("\nMean")
print(df.mean())
```

Median

```python
print("\nMedian")
print(df.median())
```

Mode

```python
print("\nMode")
print(df.mode())
```

Geometric Mean

```python
print("\nGeometric Mean")

for column in df.columns:
    print(column, ":", gmean(df[column]))
```

Harmonic Mean

```python
print("\nHarmonic Mean")

for column in df.columns:
    print(column, ":", hmean(df[column]))
```

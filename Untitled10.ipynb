```python
gender=df['Sex']
embar=df['Embarked']

count=0
total=0
for s,e in zip(gender,embar):
  if pd.isna(s) or pd.isna(e):
    continue
  total+=1
  if(s!=e):
    count+=1
print(count)
print(total)
print(count/total)
```

```python
temp=df[['Age','Fare']].dropna()

age=temp.iloc[0]['Age']
fare=temp.iloc[0]['Fare']

age2=temp.iloc[1]['Age']
fare2=temp.iloc[1]['Fare']

print(age,fare)
print(age2,fare2)
```

```python
manhanttan=abs(age-age2)+abs(fare-fare2)
print(manhanttan)
```

```python
minkowski=(((age-age2)**3)+((fare-fare2)**3))**(1/3)
print(minkowski)
```

```python
A=np.array([age,fare])
b=np.array([age2,fare2])
cosine_similarity=np.dot(A,b)/(np.linalg.norm(A)*np.linalg.norm(b))
print(cosine_similarity)
```
